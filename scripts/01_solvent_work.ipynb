```python
matched_in_mix = candidates.merge(
    mix,
    left_on=[
        "SMILES_Solute",
        "SMILES_Solvent",
        "Temperature_K"
    ],
    right_on=[
        "can_smiles_solute",
        "can_smiles_solvent1",
        "T_K"
    ]
)

mix_single = mix[
    mix["can_smiles_solvent2"].isna()
    & mix["can_smiles_solvent3"].isna()
].copy()
```

```python
big_candidates = candidates.reset_index(names="big_index").copy()

big_candidates.head()
```

```python
mix_single = mix[
    mix["can_smiles_solvent2"].isna()
    & mix["can_smiles_solvent3"].isna()
].copy()

mix_single["mol_frac_solvent1"].value_counts()
```

```python
mix_check = big_candidates.merge(
    mix_single,
    left_on=[
        "SMILES_Solute",
        "SMILES_Solvent",
        "Temperature_K"
    ],
    right_on=[
        "can_smiles_solute",
        "can_smiles_solvent1",
        "T_K"
    ],
    how="left",
    indicator=True
)
mix_check["_merge"].value_counts()
```

```python
mix_check["logS_match"] = (
    mix_check["LogS(mol/L)"].notna()
    & mix_check["logS"].notna()
    & np.isclose(
        mix_check["LogS(mol/L)"],
        mix_check["logS"],
        atol=0.005,
        rtol=0
    )
)

mix_check["logXs_match"] = (
    mix_check["LogS(mole_fraction)"].notna()
    & mix_check["logXs"].notna()
    & np.isclose(
        mix_check["LogS(mole_fraction)"],
        mix_check["logXs"],
        atol=0.005,
        rtol=0
    )
)
```

```python
mix_check["condition_match"] = (
    mix_check["_merge"] == "both"
)

mix_check["value_match"] = (
    mix_check["logS_match"]
    | mix_check["logXs_match"]
)
```

```python
mix_status = (
    mix_check
    .groupby("big_index")[
        ["condition_match", "value_match"]
    ]
    .any()
)

mix_status.head()
```

```python
big_candidates = big_candidates.merge(
    mix_status,
    left_on="big_index",
    right_index=True,
    how="left"
)

big_candidates[
    ["condition_match", "value_match"]
] = big_candidates[
    ["condition_match", "value_match"]
].fillna(False)
```

```python
big_candidates["mix_status"] = np.select(
    [
        big_candidates["value_match"],
        big_candidates["condition_match"]
    ],
    [
        "already_in_mix",
        "same_condition_different_value"
    ],
    default="not_in_mix_at_this_condition"
)
```
